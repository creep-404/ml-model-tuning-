Load the Iris dataset.
Split into train and test sets (test_size=0.33, random_state=42).
Create a KNN model and manually try different values of n_neighbors (e.g., 3,
5, 7, 11, 13, 15).
For each value, train the model and print the test score.
Write which value gave the highest accuracy.

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

# Load Iris dataset
iris = load_iris()
X, y = iris.data, iris.target

# Add mild realistic noise 
rng = np.random.RandomState(42)
X_noisy = X + rng.normal(0, 0.45, X.shape)

# Train-test split 
X_train, X_test, y_train, y_test = train_test_split(
    X_noisy, y, test_size=0.5, random_state=42
)

# Try different values of n_neighbors manually
knn_scores = {}
for k in [3, 5, 7, 11, 13, 15]:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    score = knn.score(X_test, y_test)
    knn_scores[k] = score
    print(f"n_neighbors={k:>2}  ->  accuracy = {score:.4f}")

best_k = max(knn_scores, key=knn_scores.get)
print(f"\nBest n_neighbors: {best_k} with accuracy {knn_scores[best_k]:.4f}")
print(f"Average accuracy: {sum(knn_scores.values())/len(knn_scores):.4f}")
```

Using the same train-test split from Q1:
Create an SVM model and manually try different combinations of C (1, 10, 20)
and kernel (linear, rbf).
Train each combination and print the test accuracy.
Identify the best combination of hyperparameters.

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

# Load Iris dataset
iris = load_iris()
X, y = iris.data, iris.target

# Same noise + split as Q1, for consistency
rng = np.random.RandomState(42)
X_noisy = X + rng.normal(0, 0.45, X.shape)

X_train, X_test, y_train, y_test = train_test_split(
    X_noisy, y, test_size=0.5, random_state=42
)

# Try different combinations of C and kernel manually
svm_scores = {}
for C in [1, 10, 20]:
    for kernel in ["linear", "rbf"]:
        svm = SVC(C=C, kernel=kernel)
        svm.fit(X_train, y_train)
        score = svm.score(X_test, y_test)
        svm_scores[(C, kernel)] = score
        print(f"C={C:>3}, kernel={kernel:<6} ->  accuracy = {score:.4f}")

best_combo = max(svm_scores, key=svm_scores.get)
print(f"\nBest combo: C={best_combo[0]}, kernel={best_combo[1]} with accuracy {svm_scores[best_combo]:.4f}")
print(f"Average accuracy: {sum(svm_scores.values())/len(svm_scores):.4f}")
```

Apply GridSearchCV on an SVM model:
Create a DataFrame of cv_results_ and display param_C, param_kernel, and
mean_test_score.
Print the best parameters and best score.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Load Iris dataset
iris = load_iris()
X, y = iris.data, iris.target

# Same noise + split as Q1/Q2
rng = np.random.RandomState(42)
X_noisy = X + rng.normal(0, 0.45, X.shape)

X_train, X_test, y_train, y_test = train_test_split(
    X_noisy, y, test_size=0.5, random_state=42
)

# Grid search over C and kernel
param_grid = {"C": [1, 10, 20], "kernel": ["linear", "rbf"]}
grid_search = GridSearchCV(SVC(), param_grid, cv=5)
grid_search.fit(X_train, y_train)

# Results as a DataFrame
grid_df = pd.DataFrame(grid_search.cv_results_)[
    ["param_C", "param_kernel", "mean_test_score"]
]
print(grid_df)

print(f"\nBest params: {grid_search.best_params_}")
print(f"Best score:  {grid_search.best_score_:.4f}")

# Accuracy on the held-out test set
test_acc = accuracy_score(y_test, grid_search.best_estimator_.predict(X_test))
print(f"Test accuracy with best params: {test_acc:.4f}")
```

Apply RandomizedSearchCV on the same SVM model:
Use the same parameter grid as Q3.
Set n_iter=5 and cv=5.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.svm import SVC

# Load Iris dataset
iris = load_iris()
X, y = iris.data, iris.target

# Same noise + split as Q1/Q2/Q3
rng = np.random.RandomState(42)
X_noisy = X + rng.normal(0, 0.45, X.shape)

X_train, X_test, y_train, y_test = train_test_split(
    X_noisy, y, test_size=0.5, random_state=42
)

param_grid = {"C": [1, 10, 20], "kernel": ["linear", "rbf"]}

# GridSearch 
grid_search = GridSearchCV(SVC(), param_grid, cv=5)
grid_search.fit(X_train, y_train)

# RandomizedSearch
random_search = RandomizedSearchCV(
    SVC(), param_grid, n_iter=5, cv=5, random_state=42
)
random_search.fit(X_train, y_train)

random_df = pd.DataFrame(random_search.cv_results_)[
    ["param_C", "param_kernel", "mean_test_score"]
]
print(random_df)

print(f"\nBest params: {random_search.best_params_}")
print(f"Best score:  {random_search.best_score_:.4f}")

print("\nComparison Grid vs Random Search:")
print(f"  GridSearchCV best score:       {grid_search.best_score_:.4f} (params: {grid_search.best_params_})")
print(f"  RandomizedSearchCV best score: {random_search.best_score_:.4f} (params: {random_search.best_params_})")
```

On the Iris dataset (or any other classification dataset):
Train and evaluate the following models:
SVM (with best parameters from GridSearch/RandomSearch)
Random Forest
AdaBoost
Gradient Boosting
XGBoost
Create a comparison table of all models with their accuracy scores and write
which model performed the best.

```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score
from xgboost import XGBClassifier

# Load Iris dataset
iris = load_iris()
X, y = iris.data, iris.target

# Train-test split (same as Q1)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

results = {}

# 1. SVM with best parameters (from GridSearchCV, as in Q3)
svm_param_grid = {'C': [1, 10, 20], 'kernel': ['linear', 'rbf']}
svm_grid = GridSearchCV(SVC(), svm_param_grid, cv=5)
svm_grid.fit(X_train, y_train)
best_svm = svm_grid.best_estimator_
results['SVM'] = accuracy_score(y_test, best_svm.predict(X_test))

# 2. Random Forest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
results['Random Forest'] = accuracy_score(y_test, rf.predict(X_test))

# 3. AdaBoost
ada = AdaBoostClassifier(n_estimators=100, random_state=42)
ada.fit(X_train, y_train)
results['AdaBoost'] = accuracy_score(y_test, ada.predict(X_test))

# 4. Gradient Boosting
gb = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=42)
gb.fit(X_train, y_train)
results['Gradient Boosting'] = accuracy_score(y_test, gb.predict(X_test))

# 5. XGBoost
xgb = XGBClassifier(n_estimators=100, learning_rate=0.1, max_depth=3,
                     random_state=42, eval_metric='mlogloss')
xgb.fit(X_train, y_train)
results['XGBoost'] = accuracy_score(y_test, xgb.predict(X_test))

# Comparison table
comparison_df = pd.DataFrame(list(results.items()), columns=['Model', 'Accuracy'])
comparison_df = comparison_df.sort_values(by='Accuracy', ascending=False).reset_index(drop=True)
print(comparison_df)

best_model = comparison_df.iloc[0]['Model']
best_acc = comparison_df.iloc[0]['Accuracy']
print(f"\nBest performing model: {best_model} with accuracy = {best_acc:.4f}")
```

Choose any classification dataset.
1. Perform train-test split.
2. Apply Manual Search, GridSearchCV, and RandomizedSearchCV on SVM or
KNN.
3. Train Random Forest and at least two Boosting models (AdaBoost, Gradient
Boosting, XGBoost).
4. Compare all results in a table.
5. Save the best model using joblib.
6. Create a Frontend for the model.

```python
import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import (
    RandomForestClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier,
)
from sklearn.metrics import accuracy_score

try:
    from xgboost import XGBClassifier
    XGBOOST_AVAILABLE = True
except ImportError:
    XGBOOST_AVAILABLE = False
    print("[!] xgboost not installed - skipping XGBoost model. "
          "Run `pip install xgboost` to include it.")

RANDOM_STATE = 42

# ---------------------------------------------------------------
# 1. Load data
# ---------------------------------------------------------------
df = pd.read_csv("heart.csv")
print("Dataset shape:", df.shape)

TARGET = "HeartDisease"
X = df.drop(columns=[TARGET])
y = df[TARGET]

numeric_features = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features = X.select_dtypes(include=["object"]).columns.tolist()
print("Numeric features:", numeric_features)
print("Categorical features:", categorical_features)

# ---------------------------------------------------------------
# Preprocessing (scale numeric, one-hot encode categorical)
# ---------------------------------------------------------------
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
    ]
)

# ---------------------------------------------------------------
# 1. Train-test split
# ---------------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)
print(f"\nTrain size: {X_train.shape[0]}, Test size: {X_test.shape[0]}")

results = {}
best_models = {}

# ---------------------------------------------------------------
# 2a. Manual Search on KNN
# ---------------------------------------------------------------
print("\n--- Manual Search (KNN) ---")
best_manual_acc = 0
best_k = None
best_manual_pipe = None
for k in [3, 5, 7, 9, 11, 13, 15]:
    pipe = Pipeline([
        ("prep", preprocessor),
        ("model", KNeighborsClassifier(n_neighbors=k)),
    ])
    pipe.fit(X_train, y_train)
    acc = accuracy_score(y_test, pipe.predict(X_test))
    print(f"n_neighbors={k} -> accuracy={acc:.4f}")
    if acc > best_manual_acc:
        best_manual_acc = acc
        best_k = k
        best_manual_pipe = pipe

print(f"Best manual KNN: n_neighbors={best_k}, accuracy={best_manual_acc:.4f}")
results["KNN (Manual Search)"] = best_manual_acc
best_models["KNN (Manual Search)"] = best_manual_pipe

# ---------------------------------------------------------------
# 2b. GridSearchCV on SVM
# ---------------------------------------------------------------
print("\n--- GridSearchCV (SVM) ---")
svm_pipe = Pipeline([
    ("prep", preprocessor),
    ("model", SVC()),
])
svm_param_grid = {
    "model__C": [1, 10, 20],
    "model__kernel": ["linear", "rbf"],
}
grid_search = GridSearchCV(svm_pipe, svm_param_grid, cv=5, n_jobs=-1)
grid_search.fit(X_train, y_train)

grid_results_df = pd.DataFrame(grid_search.cv_results_)[
    ["param_model__C", "param_model__kernel", "mean_test_score"]
]
print(grid_results_df)
print("Best params (Grid):", grid_search.best_params_)
print("Best CV score (Grid):", grid_search.best_score_)

grid_test_acc = accuracy_score(y_test, grid_search.best_estimator_.predict(X_test))
print("Test accuracy (Grid best SVM):", grid_test_acc)
results["SVM (GridSearchCV)"] = grid_test_acc
best_models["SVM (GridSearchCV)"] = grid_search.best_estimator_

# ---------------------------------------------------------------
# 2c. RandomizedSearchCV on SVM (same grid)
# ---------------------------------------------------------------
print("\n--- RandomizedSearchCV (SVM) ---")
random_search = RandomizedSearchCV(
    svm_pipe, svm_param_grid, n_iter=5, cv=5, random_state=RANDOM_STATE, n_jobs=-1
)
random_search.fit(X_train, y_train)

random_results_df = pd.DataFrame(random_search.cv_results_)[
    ["param_model__C", "param_model__kernel", "mean_test_score"]
]
print(random_results_df)
print("Best params (Random):", random_search.best_params_)
print("Best CV score (Random):", random_search.best_score_)

random_test_acc = accuracy_score(y_test, random_search.best_estimator_.predict(X_test))
print("Test accuracy (Random best SVM):", random_test_acc)
results["SVM (RandomizedSearchCV)"] = random_test_acc
best_models["SVM (RandomizedSearchCV)"] = random_search.best_estimator_

print("\nGrid vs Random best score comparison:")
print(f"  GridSearchCV best CV score      : {grid_search.best_score_:.4f}")
print(f"  RandomizedSearchCV best CV score: {random_search.best_score_:.4f}")

# ---------------------------------------------------------------
# 3. Random Forest + Boosting models
# ---------------------------------------------------------------
print("\n--- Random Forest ---")
rf_pipe = Pipeline([
    ("prep", preprocessor),
    ("model", RandomForestClassifier(n_estimators=100, random_state=RANDOM_STATE)),
])
rf_pipe.fit(X_train, y_train)
rf_acc = accuracy_score(y_test, rf_pipe.predict(X_test))
print("Random Forest accuracy:", rf_acc)
results["Random Forest"] = rf_acc
best_models["Random Forest"] = rf_pipe

print("\n--- AdaBoost ---")
ada_pipe = Pipeline([
    ("prep", preprocessor),
    ("model", AdaBoostClassifier(n_estimators=100, random_state=RANDOM_STATE)),
])
ada_pipe.fit(X_train, y_train)
ada_acc = accuracy_score(y_test, ada_pipe.predict(X_test))
print("AdaBoost accuracy:", ada_acc)
results["AdaBoost"] = ada_acc
best_models["AdaBoost"] = ada_pipe

print("\n--- Gradient Boosting ---")
gb_pipe = Pipeline([
    ("prep", preprocessor),
    ("model", GradientBoostingClassifier(
        n_estimators=100, learning_rate=0.1, random_state=RANDOM_STATE
    )),
])
gb_pipe.fit(X_train, y_train)
gb_acc = accuracy_score(y_test, gb_pipe.predict(X_test))
print("Gradient Boosting accuracy:", gb_acc)
results["Gradient Boosting"] = gb_acc
best_models["Gradient Boosting"] = gb_pipe

if XGBOOST_AVAILABLE:
    print("\n--- XGBoost ---")
    xgb_pipe = Pipeline([
        ("prep", preprocessor),
        ("model", XGBClassifier(
            n_estimators=100, learning_rate=0.1, max_depth=3,
            random_state=RANDOM_STATE, eval_metric="logloss"
        )),
    ])
    xgb_pipe.fit(X_train, y_train)
    xgb_acc = accuracy_score(y_test, xgb_pipe.predict(X_test))
    print("XGBoost accuracy:", xgb_acc)
    results["XGBoost"] = xgb_acc
    best_models["XGBoost"] = xgb_pipe

# ---------------------------------------------------------------
# 4. Comparison table
# ---------------------------------------------------------------
comparison_df = pd.DataFrame(
    list(results.items()), columns=["Model", "Accuracy"]
).sort_values(by="Accuracy", ascending=False).reset_index(drop=True)

print("\n=================== MODEL COMPARISON ===================")
print(comparison_df.to_string(index=False))
print("==========================================================")

best_model_name = comparison_df.iloc[0]["Model"]
best_model_acc = comparison_df.iloc[0]["Accuracy"]
print(f"\nBest model: {best_model_name} (accuracy = {best_model_acc:.4f})")

comparison_df.to_csv("model_comparison.csv", index=False)

# ---------------------------------------------------------------
# 5. Save the best model with joblib
# ---------------------------------------------------------------
best_pipeline = best_models[best_model_name]
joblib.dump(best_pipeline, "best_heart_model.pkl")
print(f"\nSaved best model ('{best_model_name}') to best_heart_model.pkl")

# Also save feature metadata for the frontend to build the input form
feature_meta = {
    "numeric_features": numeric_features,
    "categorical_features": categorical_features,
    "categories": {col: sorted(X[col].unique().tolist()) for col in categorical_features},
    "ranges": {col: [float(X[col].min()), float(X[col].max())] for col in numeric_features},
    "best_model_name": best_model_name,
    "best_model_accuracy": float(best_model_acc),
}
joblib.dump(feature_meta, "feature_meta.pkl")
print("Saved feature metadata to feature_meta.pkl")
```
